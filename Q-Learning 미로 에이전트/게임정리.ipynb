### 1. 게임 전체 구조
- 10 * 10 격자형 미로 게임
- 시작점 S에서 목표지점 G까지 이동하는 미로 게임
- 벽을 피하면서 이동해야 함(통과 안됨)

### 2. 행동 규칙
- 0: 위로 이동 = 행-1
- 1: 아래로 이동 = 행+1
- 2: 왼쪽으로 이동 = 열-1
- 3: 오른쪽으로 이동 = 열+1
```
    예를 들어 시작 위치가 (1, 1)이라면
    오른쪽이로 이동 시 (1, 2)로 이동함
```

### 3. 게임 규칙
- 일반 길로 이동 = -1 (게임 계속 진행)
- 벽에 부딪힘 = -5 (게임 계속 진행)
- 목표 지점 도착 = +100 (게임 종료)
> 에이전트가 벽에 부딪히지 않고 가능한 적은 이동으로 목표지점에 도착하는 행동 학습하는것.
